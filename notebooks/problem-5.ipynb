```python
import networkx as nx
import matplotlib.pyplot as plt
import random

chi = 4
delta = 7

def construct_graph():
    G = nx.complete_graph(chi)
    
    current_degree = chi - 1  
    extra_vertices = max(0, delta - current_degree)
    
    for i in range(extra_vertices):
        new_vertex = len(G)
        G.add_node(new_vertex)
        G.add_edge(0, new_vertex)
    
    # Greedy coloring
    coloring = nx.coloring.greedy_color(G, strategy="largest_first")
    num_colors_used = len(set(coloring.values()))
    max_degree = max(dict(G.degree()).values())

    # Print the coloring and properties of the graph
    print(f"Generated Graph: χ(G) = {num_colors_used}, Δ(G) = {max_degree}")
    print("Node Coloring:")
    for node, color in coloring.items():
        print(f"Node {node}: Color {color}")

    # Visualization of the constructed graph with coloring
    plt.figure(figsize=(8, 8))
    pos = nx.spring_layout(G)
    color_list = [coloring[node] for node in G.nodes()]
    nx.draw(G, pos, with_labels=True, node_color=color_list, node_size=700, font_size=12, cmap=plt.colormaps.get_cmap("tab20"))
    plt.title(f"Graph with χ(G) = {num_colors_used}, Δ(G) = {max_degree}")
    plt.show()

construct_graph()
```
